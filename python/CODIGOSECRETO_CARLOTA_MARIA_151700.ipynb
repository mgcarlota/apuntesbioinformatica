```python
def codigo_secreto_juego():
  codigo_secreto = "6832"
  intentos_maximos = 5
  print("BIENVENIDO AL JUEGO🤩")
  print("Tu tarea es descubrir el código secreto de 4 cifras numéricas para poder ganar🧐")
  print("Tienes 5 intentos, ¡BUENA SUERTE!🍀")
  #sirve para ir sumando intentos con cada participacion, se pone el +1 pq range nunca cuenta el ultimo numero (se quedaria en 4 intentos)
  for intento in range(1, intentos_maximos + 1):
    #la f" es para meter variables en una frase
    print(f"Intento {intento} de {intentos_maximos}")
    #es para que el juagador pueda escribir y se quede guardado en el programa dentro de la variable entrada (solo guarda texto no nº)
    entrada = input("Introduce 4 numeros aleatorios:")
    #se pone antes de todo para que en caso de que se acierte nada de lo demas se cumnpla. el break es para salir del bucle
    if entrada == "6832":
      print("¡HAS GANADO!🥳🙌🏻🏅")
      break
    else:
      #si estamos en el primer intento muestra esta pista
      if intento == 1:
        print("PISTA: la suma total del codigo es 19")
      #si lo de antes no se cumplio, comprueba esto otro (que estamos en el intento 2 y por tanto hay q enseñar esta otra pista)
      elif intento == 2:
        print("PISTA: la primera cifra es par")
      elif intento == 3:
       print("PISTA: la ultima cifra es par")
      elif intento == 4:
        print("PISTA: el segundo y tercer digito son uno par y otro impar")
    #si la entrada NO tiene solo nº di que hay un error
    if not entrada.isdigit():
      print("ERROR ❌. SOLO PUEDES INTRODUCIR NUMEROS")
    #si el codigo NO es igual a 4 caracteres de largo hay un error
    if len(entrada) != 4:
      print("ERROR ❌. DEBES INTRODUCIR 4 digitos")
    else:
      #para comprobar las posiciones del texto que es el codigo
      for posicion in range(4):
        #si el numero esta en el codigo y ademas en la misma posicion
        if entrada[posicion] == codigo_secreto[posicion]:
          print(entrada[posicion], "esta en la posicion correcta")
        #si el numero esta en el codigo pero no en la posicion correcta
        elif entrada[posicion] in codigo_secreto:
          print(entrada[posicion], "esta presente pero en la posicion incorrecta")
        #si el numero direcatmente no esta en le codigo
        else:
          print(entrada[posicion], "no esta en el codigo secreto")
    print(f"Te quedan {intentos_maximos - intento} intentos")
  #para cuando el jugador se ha quedado sin intentos y por tanto ha perdido
  else:
    intentos_maximos ==0
    print("Has perdido ☹️. El codigo secreto era", codigo_secreto)
```

```python
codigo_secreto_juego()
```
